```text
User Profile
      +
User Preferences
      +
User Vibes
      +
Wardrobe
      +
Events
      +
Weather
      +
Venue Intelligence
      +
ai_conversations.context
      +
ai_messages
```

This allows the Fashion Bot to provide increasingly personalized recommendations.

For example:

```text
User:
"I don't like bright colors."

AI:
Stores/uses this preference through the user's preferences.

Later:

User:
"What should I wear to dinner?"

AI:
Avoids recommending bright-colored outfits.
```

# 12.12 AI Security Rules

Every AI conversation must belong to the authenticated user.

The backend must always verify:

```text
conversation.user_id === authenticatedUser.id
```

For messages, the backend should verify both:

```text
message.conversation_id === requestedConversationId
AND
conversation.user_id === authenticatedUser.id
```

A user must never be able to access another user's:

- AI conversations
- AI messages
- Fashion preferences
- Wardrobe information
- AI-generated context

# 13.6 Virtual Try-On Processing Flow

The complete process works as follows:

```text
User
  ↓
Select Avatar
  ↓
Select Outfit
  ↓
POST /virtual-try-on
  ↓
Validate User
  ↓
Validate Avatar Ownership
  ↓
Validate Outfit Ownership
  ↓
Create Virtual Try-On Record
  ↓
status = processing
  ↓
Send Avatar + Outfit Data
  ↓
AI Virtual Try-On Model
  ↓
Generate Result Image
  ↓
Upload Result to Storage
  ↓
Save Result URL + Storage Key
  ↓
status = completed
  ↓
Return Result to User
```

---

# 13.7 Virtual Try-On Status

The `status` field represents the current state of the try-on request.

| Status | Description |
|---|---|
| processing | AI processing is currently running |
| completed | Result image has been successfully generated |
| failed | Processing failed |

Example:

```json
{
  "status": "processing"
}
```

After successful processing:

```json
{
  "status": "completed",
  "resultImageUrl": "https://storage.example.com/tryon/result.jpg"
}
```

If processing fails:

```json
{
  "status": "failed"
}
```

# 13.8 Avatar Validation

Before creating a virtual try-on, the backend must verify that the selected avatar belongs to the authenticated user.

The backend should enforce:

```text
avatar.user_id === authenticatedUser.id
```

A user must never be able to use another user's avatar.

---

# 13.9 Outfit Validation

The backend must also verify that the selected outfit belongs to the authenticated user.

The backend should enforce:

```text
outfit.user_id === authenticatedUser.id
```

The backend should also verify that the outfit exists before starting the AI processing.

---

# 13.10 Virtual Try-On Configuration

The `configuration` JSON object can contain optional settings for the AI try-on process.

Example:

```json
{
  "pose": "standing",
  "background": "studio",
  "fullBody": true,
  "lighting": "natural",
  "imageQuality": "high"
}
```

The exact supported configuration values should be validated by the backend before sending them to the AI processing service.

# 13.12 Result Image Storage

The generated virtual try-on image should not be stored directly inside PostgreSQL.

Instead:

```text
AI Model
   ↓
Generated Image
   ↓
Object Storage
   ↓
Storage Key + Image URL
   ↓
PostgreSQL
```

The database stores:

```json
{
  "resultImageUrl": "https://storage.example.com/tryons/result.jpg",
  "resultStorageKey": "tryons/user_uuid/result.jpg"
}
```

The actual image is stored in object storage such as AWS S3.

User-owned database queries must always be scoped to the authenticated user's ID.

### 4. Validation

All request bodies, query parameters, and path parameters should be validated before reaching the service layer.

### 5. Consistent Responses

All APIs should follow a consistent response structure:

```json
{
  "success": true,
  "message": "Operation completed successfully",
  "data": {}
}
```

### 6. Error Handling

Errors should use appropriate HTTP status codes and consistent error responses.

### 7. Database Transactions

Operations that modify multiple related tables should use database transactions where required.

### 8. AI Isolation

AI processing should be handled inside the service layer rather than directly inside route handlers.

### 9. File Storage

Images should be stored in object storage, while database tables store URLs and storage keys.

### 10. Scalability

The architecture should allow AI services, storage services, background jobs, and external APIs to be changed independently.
